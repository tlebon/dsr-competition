```python
import pandas as pd
```

#### to run on full training set, load train.csv instead of train_month_slice.parquet, the rest of the code can stay as is

```python
train = pd.read_parquet('../data/sample/train_month_slice.parquet')
train.head()
```

```python
print(train.shape)
```

## Merge transactions

```python
transactions = pd.read_csv('../data/raw/favorita-grocery-sales-forecasting/transactions.csv')
transactions.head()
```

```python
print(transactions.shape)
```

```python
# date is of different types in the two tables
merge_trans = train.astype({'date': 'str'}).merge(transactions, how='left', on=['date', 'store_nbr'])
merge_trans
```

## Merge items

```python
items = pd.read_csv('../data/raw/favorita-grocery-sales-forecasting/items.csv')
items.head()
```

```python
print(items.shape)
```

```python
merge_trans_items = merge_trans.merge(items, how='left', on='item_nbr')
merge_trans_items
```
