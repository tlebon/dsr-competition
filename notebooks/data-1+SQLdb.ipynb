```python
oil = pd.read_csv('../data/raw/oil.csv')
oil["date"] = pd.to_datetime(oil["date"])
print(oil.info())

def validate_timeseries(oil, date_col):
    assert oil[date_col].is_monotonic_increasing, "Time series is not sorted chronologically!"
    assert oil[date_col].is_unique, "Duplicate timestamps detected!"
    assert oil.index.freq is not None, "Index has no explicit frequency defined!"
    assert not oil.isna().any().any(), "Missing values exist in dataset!"
    print("Dataset is ready for time series decomposition.")
```

```python
import sqlite3

# Connect to SQLite and create a database
conn = sqlite3.connect("mini_comp.db")

# Save datasets as tables
# index = False so the row numbers aren't saved as a column, and if it exists, overwrite it.
oil.to_sql("oil", conn, if_exists="replace", index=False)
holi.to_sql("holidays", conn, if_exists="replace", index=False)
items.to_sql("item", conn, if_exists="replace", index=False)
stores.to_sql("store", conn, if_exists="replace", index=False)
transactions.to_sql("transaction", conn, if_exists="replace", index=False)

# 4. Commit changes and close connection
conn.commit()
conn.close()
```
